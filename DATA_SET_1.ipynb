```python
import pandas as pd
import matplotlib.pyplot as plt
data = pd.read_csv('/content/Titanic.csv')
gender=data['Gender'].value_counts()
plt.bar((gender.index),(gender.values))
plt.title("Male and Female Passengers")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
data = pd.read_csv('/content/Titanic.csv')
a= data["PassengerId"]
b= data["Fare"]
plt.plot(a,b)
plt.title("Fare Change Across Passengers")
plt.xlabel("Passenger ID")
plt.ylabel("Fare")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
a=data['Age']
plt.hist(a)
plt.title("Age Distribution of Passengers")
plt.xlabel("Gender")
plt.ylabel("Age")
plt.show()
```

```python
p=data["Embarked"].value_counts()
plt.pie(p.values,labels=p.index)
plt.title("Passengers Boarded")
plt.xlabel("Passenger")
plt.grid()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
a=data['Age']
b=data['Fare']
plt.scatter(a,b)
plt.title("Relationship between Age and Fare")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
data = pd.read_csv('/content/Titanic.csv')
pclass=data["Pclass"].value_counts()
plt.barh(pclass.index, pclass.values)
plt.title("Passenger Class Distribution")
plt.xlabel("Number of Passengers")
plt.ylabel("Passenger Class")
plt.show()
```
