```python
print("Rock Paper Scissor Game")
import random

while True:
        user_choice = int(input("Enter your choice: Type 0 for Rock, 1 for Paper, 2 for Scissors (3 to Quit): "))

        if user_choice == 3:
            print("Game Over. Thanks for playing!")
            break

        if user_choice < 0 or user_choice > 2:
            print("Invalid number! Please choose 0, 1, or 2.")
            continue # Restart the loop for a valid choice

        computer_choice = random.randint(0, 2)
        choices = ["Rock", "Paper", "Scissors"]
        print(f"Computer chose: {choices[computer_choice]}")

        if computer_choice == user_choice:
            print("It's a Draw.")
        elif (computer_choice == 0 and user_choice == 2) or \
             (computer_choice == 1 and user_choice == 0) or \
             (computer_choice == 2 and user_choice == 1):
            print("You Lose.")
        else:
            print("You Win!")
```
